## Feature Selection

To improve model generalization and remove non-informative attributes, 
several columns were excluded from the dataset. These features were either unique identifiers
 URLs, location-specific information, or unstructured text that were not relevant for this tabular machine learning project.

| Column | Reason for Removal |
|---------|--------------------|
| `id` | Unique identifier with no predictive value. |
| `url` | Listing URL; not useful for prediction. |
| `region_url` | Region webpage URL; irrelevant for ML. |
| `image_url` | Image link not used in tabular ML. |
| `VIN` | Unique vehicle identifier; high-cardinality feature. |
| `description` | Free-text column requiring NLP; excluded from this project. |
| `state` | Removed to build a location-independent model. |
| `region` | Location-specific feature removed for better generalization. |
| `lat` | Geographic coordinate removed to avoid location bias. |
| `long` | Geographic coordinate removed to avoid location bias. |
| `posting_date` | Listing timestamp not directly related to vehicle characteristics. |

```python
df.drop(columns=["id","url","region_url","VIN","image_url","description","county","state","lat","long","posting_date","region"],inplace=True)
```

## Feature Renaming

To improve readability, consistency, and maintainability, several feature names were renamed to more descriptive names. This naming convention makes the dataset easier to understand and improves the overall quality of the machine learning pipeline.

| Original Feature | Renamed Feature | Reason |
|------------------|-----------------|--------|
| `fuel` | `fuel_type` | Clearly specifies the type of fuel used by the vehicle. |
| `odometer` | `distance_driven_km` | Indicates the total distance travelled by the vehicle in kilometers. |
| `title_status` | `vehicle_title_status` | Provides a more descriptive name for the vehicle's legal ownership status. |
| `transmission` | `transmission_type` | Clearly indicates the type of transmission (Automatic or Manual). |
| `drive` | `drive_type` | Clarifies the drivetrain configuration (FWD, RWD, 4WD, AWD). |
| `size` | `vehicle_size` | Explicitly represents the vehicle's size category. |
| `type` | `vehicle_type` | Avoids confusion with Python's built-in `type()` function and clearly represents the vehicle category. |
| `cylinders` | `number_of_cylinders` | Clearly indicates the total number of engine cylinders in the vehicle. |

```python
df.rename(columns={
    "fuel": "fuel_type",
    "odometer": "distance_driven_km",
    "title_status": "vehicle_title_status",
    "drive": "drive_type",
    "size": "vehicle_size",
    "type": "vehicle_type",
    "cylinders": "number_of_cylinders",
    "transmission": "transmission_type"
}, inplace=True)
```

## Target Leakage Assessment

To ensure the model learns genuine relationships rather than relying on information unavailable during real-world prediction, a target leakage assessment was performed.

The assessment included:

- Verifying that all retained features are available before the selling price is known.
- Removing unique identifiers, URLs, image links, and other non-predictive attributes.
- Reviewing feature definitions to ensure no variable directly or indirectly reveals the target.

Based on these checks, no obvious source of target leakage was identified in the final feature set.
