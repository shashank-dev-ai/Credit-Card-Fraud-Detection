```python
class Solution:
    def countPrimes(self, n: int) -> int:
        if n==0 or n==1:
            return 0
        else:
            count_prime=0
            for i in range(1,n):
                if i==1:
                    continue
                else:
                    sqrt_i=int(i**0.5)
                    count_digit=0
                    for j in range(1,sqrt_i+1):
                        if i%j==0:
                            count_digit+=1
                            k=i//j
                            if k!=j:
                                count_digit=count_digit+1
                    if count_digit==2:
                        count_prime+=1
            return count_prime


```
